## Audio Information

```python
def print_audio_information(name, data):
    signal, sr = librosa.load(data)
    sample_duration = 1 / sr 
    duration = sample_duration * len(signal)
    print(f"name: {name}")
    print(f"number of samples in signal: {signal.size}")
    print(f"sampling rate: {sr} Hz")
    print(f"duration of 1 sample: {sample_duration:.6f} seconds")
    print(f"duration of audio signal: {duration:.6f} seconds\n")
```

## Audio Augmentation 

### White Noise Augmentation

```python
def add_white_noise(signal, noise_percentage_factor):
    noise = np.random.normal(0, signal.std(), signal.size)
    augmented_signal = signal + noise * noise_percentage_factor
    return augmented_signal
    
```

## Time Stretching Augmentation

```python
def time_stretch(signal, time_stretch_rate):
    """Time stretching implemented with librosa:
    https://librosa.org/doc/main/generated/librosa.effects.pitch_shift.html?highlight=pitch%20shift#librosa.effects.pitch_shift
    """
    return librosa.effects.time_stretch(y=signal, rate=time_stretch_rate)
```

## Pitch Scaling Augmentation

```python
def pitch_scale(signal, sr, num_semitones):
    """Pitch scaling implemented with librosa:
    https://librosa.org/doc/main/generated/librosa.effects.pitch_shift.html?highlight=pitch%20shift#librosa.effects.pitch_shift
    """
    return librosa.effects.pitch_shift(y=signal, sr=sr, n_steps=num_semitones)
```

### Polarity Inversion Augmentation

```python
def invert_polarity(signal):
    return signal * -1
```

### Random Gain Augmentation

```python
def random_gain(signal, min_gain_factor, max_gain_factor):
    gain_factor = random.uniform(min_gain_factor, max_gain_factor)
    return signal * gain_factor
```

## Visualizing Data Augmentation

```python
def plot_signal_and_augmented_signal(signal, augmented_signal, sr):
    fig, ax = plt.subplots(nrows=2)
    librosa.display.waveshow(signal, sr=sr, ax=ax[0])
    ax[0].set(title="Original signal")
    librosa.display.waveshow(augmented_signal, sr=sr, ax=ax[1])
    ax[1].set(title="Augmented signal")
    plt.show()
```

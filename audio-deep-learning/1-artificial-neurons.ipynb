## Artificial Neurons

```python
def sigmoid(x):
    return 1.0 / (1 + math.exp(-x))
```

```python
def activate(inputs, weights):
    # net input 
    h = 0
    
    for x, w in zip(inputs, weights):
        h += x * w
        
    # activation 
    return sigmoid(h)
```

```python
inputs = [0.5, 0.3, 0.2]
weights=[0.4, .7, 0.2]
output = activate(inputs, weights)
print(f"output: {output}")
```
